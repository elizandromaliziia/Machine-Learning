```python
#Passo 1: Importar as bibliotecas
from sklearn.datasets import load_iris
import pandas as pd
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
```

```python
#Passo 2: Carregar e visualizar os dados
#2.1: Carregar os dados
iris = load_iris()

#2.2: Visualizar o que ha dentro do conjunto
print(f'''Nomes das features (medidas): {iris.feature_names}
Nomes das classes (espécies): {iris.target_names}''')
```

```python
#Passo 3: Transformar em DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target
df.head()
```

```python
from re import X
#Passo 3: Separar Dados e as Respostas
x = df.drop('target', axis=1)
y = df['target']
```

```python
#Passo 4: Divisão dos dados de treino e de teste
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42)

print(f'''Total de amostras: {len(x)}
          Amostras para treino: {len(x_train)}
          Amostras para treino: {len(x_test)}''')
```

```python
#Passo 5: Criando e treinando o modelo
#5.1 Carregar o modelo (cru)
model = KNeighborsClassifier(n_neighbors=3)

#5.2 Treinar o modelo
model.fit(x_train, y_train)

print('Modelo Treinado com sucesso!')
```

```python
#Passo 6:Fazendo a previsão
#6.1 Fazer as previsões com os dados de teste
previsao = model.predict(x_test)

#6.2 Analisar algumas previsões
print(f'Previsoe do modelo: {previsao[:5]}')
print(f'Respostas reais: \n{y_test[:5]}')
```

```python
#Passo 7: Validando a acurácia
#7.1: Avaliar a performance
acuracia = accuracy_score(y_test, previsao)
print(f'Acuracia do modelo: {acuracia:.2%}')
```
